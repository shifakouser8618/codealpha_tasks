```python
import pandas as pd
import numpy as np

# 1. Load the Dataset
df = pd.read_csv("wikipedia_scraped_data.csv")

# 2. Explore Data Structure & Types
print("=== DATASET OVERVIEW ===")
print(f"Shape of Dataset: {df.shape[0]} rows, {df.shape[1]} columns\n")

print("=== DATA TYPES & NON-NULL COUNTS ===")
print(df.info())
print("\n")

# 3. Detect Missing Values & Duplicates
print("=== MISSING VALUES PER COLUMN ===")
print(df.isnull().sum())
print("\n")

print("=== DUPLICATE RECORDS ===")
duplicate_count = df.duplicated().sum()
print(f"Total duplicate rows found: {duplicate_count}\n")

# 4. Clean Data (Handle Duplicates & Missing Values)
df_clean = df.drop_duplicates().copy()

# Feature Engineering: Add string length metrics for exploratory insights
if 'Article Title' in df_clean.columns:
    df_clean['Title_Length'] = df_clean['Article Title'].astype(str).apply(len)
    df_clean['Word_Count'] = df_clean['Article Title'].astype(str).apply(lambda x: len(x.split()))

# 5. Summary Statistics & Hypotheses Validation
print("=== NUMERICAL SUMMARY STATISTICS ===")
print(df_clean.describe())
print("\n")

print("=== CATEGORICAL SUMMARY (Top 5 Entries) ===")
print(df_clean.describe(include=['O']))
print("\n")

# 6. Save Cleaned & Analyzed Data
df_clean.to_csv("eda_cleaned_wikipedia_data.csv", index=False)
print("EDA Complete! Cleaned file saved as 'eda_cleaned_wikipedia_data.csv'.")
```
