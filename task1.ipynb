```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# Target URL (A public sandbox site designed for scraping practice)
URL = "http://books.toscrape.com/"

def scrape_books(url):
    # Send an HTTP GET request
    response = requests.get(url)
    
    # Check if request was successful
    if response.status_code != 200:
        print(f"Failed to retrieve page, status code: {response.status_code}")
        return None

    # Parse HTML content
    soup = BeautifulSoup(response.text, "html.parser")
    
    books_data = []
    
    # Locate all book containers
    articles = soup.find_all("article", class_="product_pod")
    
    for article in articles:
        # Extract Title
        title = article.h3.a["title"]
        
        # Extract Price
        price = article.find("p", class_="price_color").text.strip()
        
        # Extract Availability
        availability = article.find("p", class_="instock availability").text.strip()
        
        books_data.append({
            "Title": title,
            "Price": price,
            "Availability": availability
        })
        
    return pd.DataFrame(books_data)

# Run the scraper
df_scraped = scrape_books(URL)

# Save the dataset locally for Task 2 (EDA)
df_scraped.to_csv("scraped_books_data.csv", index=False)

print("Scraping completed successfully! Preview of the scraped dataset:")
print(df_scraped.head())
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# Target URL provided
URL = "https://en.wikipedia.org/wiki/Main_Page"

def scrape_wikipedia_main():
    # Wikipedia requires a descriptive User-Agent header to identify your bot/script
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) EducationalScraper/1.0 (student project)"
    }
    
    response = requests.get(URL, headers=headers)
    
    if response.status_code != 200:
        print(f"Failed to fetch page. Status code: {response.status_code}")
        return None

    soup = BeautifulSoup(response.text, "html.parser")
    
    scraped_items = []
    
    # Example: Extracting "Today's featured article" heading and summary from Wikipedia's Main Page
    # Wikipedia main page uses specific container IDs/classes
    featured_article_box = soup.find("div", {"id": "mp-tfa"})
    
    if featured_article_box:
        # Find all links within the featured article section
        links = featured_article_box.find_all("a")
        
        for link in links:
            title = link.get("title")
            href = link.get("href")
            if title and href:
                scraped_items.append({
                    "Article Title": title,
                    "Wikipedia Link": f"https://en.wikipedia.org{href}"
                })
                
    # Convert to DataFrame
    df = pd.DataFrame(scraped_items).drop_duplicates()
    return df

# Run the scraper
df_wiki = scrape_wikipedia_main()

if df_wiki is not None and not df_wiki.empty:
    print("Scraping successful! Here are some related article links extracted from Today's Featured Article:")
    print(df_wiki.head(10))
    
    # Save to CSV for Task 2 (EDA)
    df_wiki.to_csv("wikipedia_scraped_data.csv", index=False)
else:
    print("Could not extract items. The page structure might have shifted.")
```
