```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load EDA dataset
df = pd.read_csv("eda_cleaned_wikipedia_data.csv")

# Set aesthetic theme
sns.set_theme(style="whitegrid", palette="muted")
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot 1: Bar Chart of Word Counts
sns.countplot(x='Word_Count', data=df, ax=axes[0], palette="Blues_d")
axes[0].set_title("Frequency of Word Counts in Article Titles", fontsize=12, fontweight='bold')
axes[0].set_xlabel("Word Count per Title")
axes[0].set_ylabel("Number of Articles")

# Plot 2: Boxplot showing spread & potential anomalies
sns.boxplot(x=df['Title_Length'], ax=axes[1], color="mediumseagreen")
axes[1].set_title("Spread and Outliers of Title Lengths", fontsize=12, fontweight='bold')
axes[1].set_xlabel("Character Length")

plt.tight_layout()
plt.savefig("task3_data_visualizations.png", dpi=300)
plt.show()

print("Visualizations saved successfully as 'task3_data_visualizations.png'!")
```
